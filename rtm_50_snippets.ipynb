## 01 — Square to Circle Morph
**What this does:**
    A blue square smoothly morphs into a yellow circle.
**Key ideas:**
    - Transform() interpolates the points of one shape into the points of another.
    - Only ONE object is ever on stage: the square *becomes* the circle.
**Try changing:**
    - run_time=2.0 -> 5.0 for a slower morph.
    - Morph a triangle into a pentagon instead (see snippet 27).

````python
import os

from real_time_manim import *  # Vulkan renderer + full ManimCE namespace
from real_time_manim.record import fast_record_scene  # hidden-window MP4 recorder

OUTPUT = os.path.join("outputs", "01_square_to_circle.mp4")
os.makedirs("outputs", exist_ok=True)


class SquareToCircle(Scene):
    def construct(self):
        win = MLWindow(1920, 1080)  # the real-time Vulkan window (fixed 1080p60)
        win.scene = self            # REQUIRED: bind this scene to the window

        square = Square(side_length=2.5).set_fill(BLUE, 0.6).set_stroke(WHITE, 3)
        circle = Circle(radius=1.8).set_fill(YELLOW, 0.6).set_stroke(WHITE, 3)

        self.add(square)                              # put the square on stage
        win.play(Transform(square, circle), run_time=2.0)  # morph square -> circle
        win.play(Wait(1.0))

        win.close()  # REQUIRED: flush frames and finish the MP4


from IPython.display import Video, display

result = fast_record_scene(SquareToCircle, OUTPUT, overwrite=True)
display(Video(OUTPUT, width=854))
````

## 10 — A Staggered Row of Squares
**What this does:**
    Eight squares fade in one after another, left to right, like dominoes.
**Key ideas:**
    - VGroup(*squares) bundles mobjects into one container you can treat as one.
    - .arrange(RIGHT, buff=0.3) lines the children up with 0.3 units of gap.
    - LaggedStart(anim1, anim2, ..., lag_ratio=0.15) starts each animation a
      fraction of a second after the previous one.
**Try changing:**
    - lag_ratio=0.15 -> 0.5 (bigger gaps between starts).
    - FadeIn -> Create in the list comprehension for a drawing wave.

````python
import os

from real_time_manim import *  # Vulkan renderer + full ManimCE namespace
from real_time_manim.record import fast_record_scene  # hidden-window MP4 recorder

OUTPUT = os.path.join("outputs", "10_staggered_row.mp4")
os.makedirs("outputs", exist_ok=True)


class StaggeredRow(Scene):
    def construct(self):
        win = MLWindow(1920, 1080)  # the real-time Vulkan window (fixed 1080p60)
        win.scene = self            # REQUIRED: bind this scene to the window

        squares = VGroup(*[Square(side_length=0.9) for _ in range(8)])
        squares.arrange(RIGHT, buff=0.3).set_fill(BLUE, 0.6).set_stroke(WHITE, 2)

        # One FadeIn per square, started with a small delay each.
        win.play(
            LaggedStart(*[FadeIn(sq) for sq in squares], lag_ratio=0.15),
            run_time=2.0,
        )
        win.play(Wait(1.0))

        win.close()  # REQUIRED: flush frames and finish the MP4


from IPython.display import Video, display

result = fast_record_scene(StaggeredRow, OUTPUT, overwrite=True)
display(Video(OUTPUT, width=854))
````

## 11 — A Short Story in One Line
**What this does:**
    One Succession() call chains three animations: a square is drawn,
    rotates 45°, then fades away.
**Key ideas:**
    - Succession(anim1, anim2, ...) plays its animations strictly one after
      another, inside a SINGLE play() call.
    - Each inner animation carries its own run_time; the play's run_time is
      the sum of the parts.
    - FadeOut() removes the mobject from the scene when it finishes.
**Try changing:**
    - Insert another step, like Rotate(square, angle=-PI/4) before the fade.
    - Add lag_ratio=0.5 to Succession so each step overlaps the next.

````python
import os

from real_time_manim import *  # Vulkan renderer + full ManimCE namespace
from real_time_manim.record import fast_record_scene  # hidden-window MP4 recorder

OUTPUT = os.path.join("outputs", "11_story_sequence.mp4")
os.makedirs("outputs", exist_ok=True)


class StorySequence(Scene):
    def construct(self):
        win = MLWindow(1920, 1080)  # the real-time Vulkan window (fixed 1080p60)
        win.scene = self            # REQUIRED: bind this scene to the window

        square = Square(side_length=2.5).set_fill(BLUE, 0.6).set_stroke(WHITE, 3)

        # Draw it, spin it, let it go — three acts, one play.
        win.play(
            Succession(
                Create(square, run_time=1.2),
                Rotate(square, angle=PI / 4, run_time=1.0),
                FadeOut(square, run_time=0.8),
            ),
            run_time=3.0,
        )

        win.close()  # REQUIRED: flush frames and finish the MP4


from IPython.display import Video, display

result = fast_record_scene(StorySequence, OUTPUT, overwrite=True)
display(Video(OUTPUT, width=854))
````

## 17 — Expand a Binomial
**What this does:**
    (a+b)² dissolves into its expansion a² + 2ab + b².
**Key ideas:**
    - FadeTransform(src, dst) is Transform plus a fade: the old formula
      dims out while the new one fades in at the same spot.
    - This is the standard "algebra move" animation: identity on the left,
      expansion on the right.
**Try changing:**
    - The two formulas (e.g. sin²x + cos²x -> 1).

## 27 — Triangle to Pentagon and Back
**What this does:**
    A triangle morphs into a pentagon, holds, then morphs back.
**Key ideas:**
    - Regular polygons are built from points on a circle:
      cos(2*pi*k/n), sin(2*pi*k/n) for k = 0..n-1.
    - Polygon(*points) takes the list of vertices and closes the loop.
    - .copy() freezes a snapshot of a mobject — Transform(tri, tri_copy)
      morphs back to the exact original shape.
    - Transform works across different vertex counts (3 vs 5).
**Try changing:**
    - n=5 -> n=8 for an octagon, or n=100 for a near-circle.
    - Make the two polygons different colors.

````python
import os

import numpy as np

from real_time_manim import *  # Vulkan renderer + full ManimCE namespace
from real_time_manim.record import fast_record_scene  # hidden-window MP4 recorder

OUTPUT = os.path.join("outputs", "27_triangle_pentagon.mp4")
os.makedirs("outputs", exist_ok=True)


def ngon(n, radius, color):
    """Build a regular n-gon from n evenly spaced points on a circle."""
    points = [
        np.array([radius * np.cos(2 * np.pi * k / n - np.pi / 2),
                  radius * np.sin(2 * np.pi * k / n - np.pi / 2), 0.0])
        for k in range(n)
    ]
    return Polygon(*points).set_fill(color, 0.5).set_stroke(WHITE, 2)


class TrianglePentagon(Scene):
    def construct(self):
        win = MLWindow(1920, 1080)  # the real-time Vulkan window (fixed 1080p60)
        win.scene = self            # REQUIRED: bind this scene to the window

        triangle = ngon(3, 1.6, BLUE)
        pentagon = ngon(5, 1.6, GOLD)
        triangle_copy = triangle.copy()  # snapshot for the return trip

        self.add(triangle)
        win.play(Transform(triangle, pentagon), run_time=2.0)      # 3 -> 5 sides
        win.play(Transform(triangle, triangle_copy), run_time=2.0)  # and back
        win.play(Wait(1.0))

        win.close()  # REQUIRED: flush frames and finish the MP4


from IPython.display import Video, display

result = fast_record_scene(TrianglePentagon, OUTPUT, overwrite=True)
display(Video(OUTPUT, width=854))
````

## 45 — A Breathing Circle
**What this does:**
    A circle inhales and exhales: its radius oscillates ±20% four times.
**Key ideas:**
    - An updater receives (mobject, dt) every frame while a play runs;
      dt is the seconds since the previous frame.
    - phase += 2*PI*dt advances exactly one cycle per second.
    - m.become(new_circle) rebuilds the geometry each frame — the radius
      is set absolutely (1.5 * (1 + 0.2*sin(phase))), so nothing drifts.
    - win.play(Wait(4.0)) keeps the clock running so the updater can work.
**Try changing:**
    - 0.2 -> 0.5 (deep breaths), 2*PI*dt -> 4*PI*dt (twice the tempo).
    - Rebuild with a different color inside become() to cycle colors too.

````python
import os

import numpy as np

from real_time_manim import *  # Vulkan renderer + full ManimCE namespace
from real_time_manim.record import fast_record_scene  # hidden-window MP4 recorder

OUTPUT = os.path.join("outputs", "45_breathing_circle.mp4")
os.makedirs("outputs", exist_ok=True)


class BreathingCircle(Scene):
    def construct(self):
        win = MLWindow(1920, 1080)  # the real-time Vulkan window (fixed 1080p60)
        win.scene = self            # REQUIRED: bind this scene to the window

        circle = Circle(radius=1.5).set_stroke(TEAL, 6)
        phase = [0.0]  # mutable box so the updater can accumulate time

        def breathe(m, dt):
            phase[0] += 2 * np.pi * dt                      # one cycle per second
            r = 1.5 * (1 + 0.2 * np.sin(phase[0]))          # 1.2 .. 1.8
            m.become(Circle(radius=r).set_stroke(TEAL, 6))  # rebuild the ring

        self.add(circle)
        circle.add_updater(breathe)
        win.play(Wait(4.0))  # the play keeps the updater clock running
        circle.clear_updaters()
        win.play(Wait(1.0))

        win.close()  # REQUIRED: flush frames and finish the MP4


from IPython.display import Video, display

result = fast_record_scene(BreathingCircle, OUTPUT, overwrite=True)
display(Video(OUTPUT, width=854))
````

## 50 — The Grand Finale
**What this does:**
    One big square dissolves into four smaller squares, then a closing
    line fades in underneath. Everything from the last 49 snippets in one
    shot: groups, fades, and text.
**Key ideas:**
    - FadeOut() removes the big square while FadeIn() brings the group of
      four on stage — a clean swap between two unrelated mobjects.
    - Each quadrant gets its own fill color for free contrast.
    - Text + FadeIn closes the show.
**Try changing:**
    - The four colors, or split into 9 squares in a 3x3 grid.
    - Put your own name in the closing text.

````python
import os

from real_time_manim import *  # Vulkan renderer + full ManimCE namespace
from real_time_manim.record import fast_record_scene  # hidden-window MP4 recorder

OUTPUT = os.path.join("outputs", "50_grand_finale.mp4")
os.makedirs("outputs", exist_ok=True)


class GrandFinale(Scene):
    def construct(self):
        win = MLWindow(1920, 1080)  # the real-time Vulkan window (fixed 1080p60)
        win.scene = self            # REQUIRED: bind this scene to the window

        big = Square(side_length=3.0).set_fill(BLUE, 0.5).set_stroke(WHITE, 3)

        quadrants = [
            (RED, [0.75, 0.75, 0.0]),     # top right
            (GOLD, [-0.75, 0.75, 0.0]),   # top left
            (TEAL, [0.75, -0.75, 0.0]),   # bottom right
            (GREEN, [-0.75, -0.75, 0.0]),  # bottom left
        ]
        small = VGroup(*[
            Square(side_length=1.5).set_fill(color, 0.6).set_stroke(WHITE, 2).move_to(pos)
            for color, pos in quadrants
        ])

        self.add(big)
        win.play(FadeOut(big), run_time=0.8)   # the big square leaves...
        win.play(FadeIn(small), run_time=1.2)  # ...and the four arrive
        win.play(Wait(0.5))

        closing = Text("50 snippets — go make something!", font_size=48).to_edge(DOWN)
        win.play(FadeIn(closing), run_time=1.5)
        win.play(Wait(1.5))

        win.close()  # REQUIRED: flush frames and finish the MP4


from IPython.display import Video, display

result = fast_record_scene(GrandFinale, OUTPUT, overwrite=True)
display(Video(OUTPUT, width=854))
````
